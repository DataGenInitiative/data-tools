```python
import pandas as pd
import os
import gc
import shutil
```

###### Data initial drop

```python
# Set the base path to where the data is stored
base_path = r'/Users/marcocassar/Projects/temp data filter/rtn_data/'
```

```python
# Drop all files that have no rows
delete_count = 0

for filename in os.listdir(base_path):
    if filename.lower().endswith(".csv"):
        filepath = os.path.join(base_path, filename)
        try:
            # Use low_memory=False to silence mixed-type warnings
            df = pd.read_csv(filepath, encoding="ISO-8859-1", low_memory=False)
            if df.shape[0] == 0:  # no rows at all
                print(f"Deleting file with no rows: {filename}")
                delete_count += 1
                os.remove(filepath)
        except Exception as e:
            print(f"Skipping {filename}, error: {e}")
        finally:
            if "df" in locals():
                del df
                gc.collect()
            print(f"Deleted a total of {delete_count} rows")
```

```python
# Create an array of all available items to go through then create functions that let you work your way through each file
data_items = os.listdir(base_path)
print(f"Total of {len(data_items)} files")

# This deletes the file at index specified | I just use default zero for evey thing and just repeatedly run the bottom three cells
def delete_index(index: int = 0):
    f_path = os.path.join(base_path, data_items[index])
    if data_items[index] != 'kept':
        print(f"Deleting file: {f_path}")
        os.remove(f_path)
    del data_items[index]

# Loads the csv at index
def load_index(index: int = 0):
    print(f"Loading file: {data_items[index]}")
    print(f"Loaded {len(df.columns - 1)} columns\n")
    print(f"Columns are {", ".join(df.columns)}\n")
    print(f"{len(data_items)} files remaining\n")
    return pd.read_csv(base_path + data_items[index], encoding="ISO-8859-1")

# Moves csv files to a subdir called kept to go through in a more detailed approach later
def keep_index(index: int = 0):
    src = os.path.join(base_path, data_items[index])
    kept_dir = os.path.join(base_path, "kept")
    os.makedirs(kept_dir, exist_ok=True)
    dst = os.path.join(kept_dir, data_items[index])

    print(f"Keeping file: {data_items[index]} -> {dst}")
    shutil.move(src, dst)

    del data_items[index]
```
